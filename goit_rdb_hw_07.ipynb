```python
import os
import sys
import pandas as pd
import pgserver

from urllib.request import urlretrieve
from sqlalchemy import create_engine, text

print("Python:", sys.version)

pg = pgserver.get_server('/tmp/hw7_pg', cleanup_mode='stop')
engine = create_engine(pg.get_uri(), future=True)

with engine.connect() as conn:
    print(conn.execute(text("SELECT version()")).scalar_one())
```

Dataset: NYC TLC Yellow Taxi Trip Records, March 2024. Source: https://www.nyc.gov/site/tlc/about/tlc-trip-record-data.page

```python
URL = (
    "https://d37ci6vzurychx.cloudfront.net/trip-data/"
    "yellow_tripdata_2024-03.parquet"
)

parquet_path = "/content/yellow_tripdata_2024-03.parquet"

urlretrieve(URL, parquet_path)

columns = [
    "tpep_pickup_datetime",
    "tpep_dropoff_datetime",
    "PULocationID",
    "DOLocationID",
    "trip_distance",
    "fare_amount",
    "passenger_count",
]

df = pd.read_parquet(parquet_path, columns=columns)

df = df[
    (df["tpep_pickup_datetime"] >= "2024-03-01") &
    (df["tpep_pickup_datetime"] < "2024-04-01")
]

df = df.sample(
    n=min(500_000, len(df)),
    random_state=42
).copy()

print("Sampled date range:")
print(df["tpep_pickup_datetime"].min())
print(df["tpep_pickup_datetime"].max())
```

```python
rows_before = len(df)

df = df[
    (df["fare_amount"] > 0) &
    (df["fare_amount"] < 500) &
    (df["trip_distance"] > 0) &
    (df["trip_distance"] < 100) &
    (df["PULocationID"].notna())
].copy()

print("Before cleaning:", rows_before)
print("After cleaning:", len(df))
print("Removed rows:", rows_before - len(df))
```

```python
df.to_sql(
    "nyc_taxi_raw",
    engine,
    if_exists="replace",
    index=False,
    chunksize=10_000,
)

pd.read_sql("""
    SELECT COUNT(*) AS n_rows
    FROM nyc_taxi_raw;
""", engine)
```

```python
with engine.begin() as conn:
    conn.execute(text("""
        DROP TABLE IF EXISTS t7_hw_taxi_daily;

        CREATE TABLE t7_hw_taxi_daily AS
        SELECT
            "PULocationID"::INTEGER AS pickup_zone_id,
            DATE_TRUNC('day', tpep_pickup_datetime::TIMESTAMP)::DATE AS ts_day,
            COUNT(*)::INTEGER AS trip_count,
            AVG(fare_amount)::NUMERIC(10, 2) AS avg_fare,
            SUM(trip_distance)::NUMERIC(12, 2) AS total_distance
        FROM nyc_taxi_raw
        WHERE fare_amount > 0
          AND fare_amount < 500
          AND trip_distance > 0
          AND trip_distance < 100
          AND "PULocationID" IS NOT NULL
        GROUP BY
            "PULocationID"::INTEGER,
            DATE_TRUNC('day', tpep_pickup_datetime::TIMESTAMP)::DATE;

        ALTER TABLE t7_hw_taxi_daily
            ADD PRIMARY KEY (pickup_zone_id, ts_day);
    """))

print("Daily aggregation created successfully.")
```

```python
with engine.begin() as conn:
    conn.execute(text("""
        DROP VIEW IF EXISTS v_t7_hw_taxi_features;
        DROP TABLE IF EXISTS t7_hw_taxi_daily_spine;

        CREATE TABLE t7_hw_taxi_daily_spine AS
        WITH zones AS (
            SELECT DISTINCT pickup_zone_id
            FROM t7_hw_taxi_daily
        ),
        calendar AS (
            SELECT generate_series(
                DATE '2024-03-01',
                DATE '2024-03-31',
                INTERVAL '1 day'
            )::DATE AS ts_day
        )
        SELECT
            z.pickup_zone_id,
            c.ts_day,
            COALESCE(d.trip_count, 0)::INTEGER AS trip_count,
            d.avg_fare,
            COALESCE(d.total_distance, 0)::NUMERIC(12, 2) AS total_distance
        FROM zones z
        CROSS JOIN calendar c
        LEFT JOIN t7_hw_taxi_daily d
            ON d.pickup_zone_id = z.pickup_zone_id
           AND d.ts_day = c.ts_day;

        ALTER TABLE t7_hw_taxi_daily_spine
            ADD PRIMARY KEY (pickup_zone_id, ts_day);
    """))

print("Calendar spine created successfully.")
```

## Завдання 2. Time-series ML features

```python
with engine.begin() as conn:
    conn.execute(text("""
        CREATE OR REPLACE VIEW v_t7_hw_taxi_features AS
        WITH features_basic AS (
            SELECT
                pickup_zone_id,
                ts_day,
                trip_count,
                avg_fare,
                total_distance,

                LAG(trip_count, 1) OVER w_pid AS trips_lag_1d,
                LAG(trip_count, 7) OVER w_pid AS trips_lag_7d,
                LAG(trip_count, 30) OVER w_pid AS trips_lag_30d,

                AVG(avg_fare) OVER w7 AS rolling_avg_fare_7d,
                COUNT(avg_fare) OVER w7 AS fare_days_7d,

                STDDEV_SAMP(avg_fare) OVER w30 AS rolling_std_fare_30d,
                COUNT(avg_fare) OVER w30 AS fare_days_30d,

                SUM(trip_count) OVER we AS cumulative_trips,
                AVG(trip_count) OVER we AS expanding_avg_trips

            FROM t7_hw_taxi_daily_spine

            WINDOW
                w_pid AS (
                    PARTITION BY pickup_zone_id
                    ORDER BY ts_day
                ),
                w7 AS (
                    PARTITION BY pickup_zone_id
                    ORDER BY ts_day::TIMESTAMP
                    RANGE BETWEEN INTERVAL '6 days' PRECEDING
                              AND CURRENT ROW
                ),
                w30 AS (
                    PARTITION BY pickup_zone_id
                    ORDER BY ts_day::TIMESTAMP
                    RANGE BETWEEN INTERVAL '29 days' PRECEDING
                              AND CURRENT ROW
                ),
                we AS (
                    PARTITION BY pickup_zone_id
                    ORDER BY ts_day
                    ROWS BETWEEN UNBOUNDED PRECEDING
                             AND CURRENT ROW
                )
        )
        SELECT *
        FROM features_basic
        WHERE trips_lag_30d IS NOT NULL;
    """))

print("Feature view created successfully.")
```

```python
display(pd.read_sql("""
    SELECT
        COUNT(*) AS n_rows,
        COUNT(DISTINCT pickup_zone_id) AS n_zones,
        MIN(ts_day) AS first_day,
        MAX(ts_day) AS last_day,
        COUNT(*) FILTER (
            WHERE rolling_avg_fare_7d IS NOT NULL
        ) AS valid_rolling_avg,
        COUNT(*) FILTER (
            WHERE rolling_std_fare_30d IS NOT NULL
        ) AS valid_rolling_std
    FROM v_t7_hw_taxi_features;
""", engine))

display(pd.read_sql("""
    SELECT *
    FROM v_t7_hw_taxi_features
    ORDER BY pickup_zone_id
    LIMIT 10;
""", engine))
```

## Завдання 3. JSONB-таблиця для подій

```python
with engine.begin() as conn:
    conn.execute(text("""
        DROP TABLE IF EXISTS t7_hw_events;

        CREATE TABLE t7_hw_events (
            event_id    INTEGER GENERATED BY DEFAULT AS IDENTITY PRIMARY KEY,
            user_id     INTEGER NOT NULL,
            event_ts    TIMESTAMPTZ NOT NULL DEFAULT NOW(),
            event_type  TEXT NOT NULL,
            payload     JSONB NOT NULL
        );

        INSERT INTO t7_hw_events (user_id, event_ts, event_type, payload)
        SELECT
            (n % 1000) + 1 AS user_id,
            TIMESTAMPTZ '2024-04-01 00:00:00+00'
                - ((n % 60) || ' days')::INTERVAL AS event_ts,

            CASE (n % 5)
                WHEN 0 THEN 'page_view'
                WHEN 1 THEN 'click'
                WHEN 2 THEN 'add_to_cart'
                WHEN 3 THEN 'checkout'
                ELSE 'recommendation_shown'
            END AS event_type,

            CASE (n % 5)
                WHEN 0 THEN jsonb_build_object(
                    'url', '/product/' || ((n % 200) + 1),
                    'referrer', (ARRAY['google','direct','email','fb'])[(n % 4) + 1],
                    'device', (ARRAY['mobile','desktop','tablet'])[(n % 3) + 1]
                )

                WHEN 1 THEN jsonb_build_object(
                    'element_id', 'btn_' || ((n % 50) + 1),
                    'page_url', '/category/' || ((n % 20) + 1),
                    'device', (ARRAY['mobile','desktop'])[(n % 2) + 1]
                )

                WHEN 2 THEN jsonb_build_object(
                    'product_id', (n % 200) + 1,
                    'price', ((n % 100) + 10)::NUMERIC(10, 2),
                    'quantity', (n % 5) + 1,
                    'tags', jsonb_build_array(
                        (ARRAY['promo','new','clearance'])[(n % 3) + 1]
                    )
                )

                WHEN 3 THEN jsonb_build_object(
                    'order_id', 'o_' || (n + 100000),
                    'total', ((n % 500) + 50)::NUMERIC(10, 2),
                    'items_count', (n % 10) + 1,
                    'payment_method', (ARRAY['card','paypal','crypto'])[(n % 3) + 1]
                )

                ELSE jsonb_build_object(
                    'model_name', 'rec_v' || ((n % 5) + 1),
                    'model_version', '1.' || (n % 10),
                    'recommendations', jsonb_build_array(
                        (n % 200) + 1,
                        ((n + 1) % 200) + 1,
                        ((n + 2) % 200) + 1
                    ),
                    'experiment', 'exp_' || ((n % 3) + 1)
                )
            END AS payload

        FROM generate_series(1, 50000) AS n;

        ANALYZE t7_hw_events;
    """))

print("JSONB event log created successfully.")
```

## Завдання 4. GIN-index demo + EXPLAIN ANALYZE

```python
with engine.begin() as conn:
    conn.execute(text("""
        DROP INDEX IF EXISTS idx_t7_events_payload_gin_pathops;
        DROP INDEX IF EXISTS idx_t7_events_payload_gin_ops;
        DROP INDEX IF EXISTS idx_t7_events_device_btree;

        ANALYZE t7_hw_events;
    """))
```

```python
plan = pd.read_sql("""
    EXPLAIN (ANALYZE, BUFFERS)
    SELECT COUNT(*)
    FROM t7_hw_events
    WHERE payload @> jsonb_build_object('payment_method', 'card');
""", engine)

print("\n".join(plan["QUERY PLAN"]))
```

```python
with engine.begin() as conn:
    conn.execute(text("""
        CREATE INDEX idx_t7_events_payload_gin_pathops
            ON t7_hw_events USING GIN (payload jsonb_path_ops);

        ANALYZE t7_hw_events;
    """))

print("GIN index created successfully.")
```

```python
display(pd.read_sql("""
    SELECT
        pg_size_pretty(
            pg_relation_size('idx_t7_events_payload_gin_pathops')
        ) AS gin_index_size;
""", engine))
```

```python
with engine.begin() as conn:
    conn.execute(text("""
        CREATE INDEX idx_t7_events_device_btree
            ON t7_hw_events ((payload ->> 'device'));

        ANALYZE t7_hw_events;
    """))

print("Expression B-tree index created successfully.")
```

```python
plan = pd.read_sql("""
    EXPLAIN (ANALYZE, BUFFERS)
    SELECT COUNT(*)
    FROM t7_hw_events
    WHERE payload ->> 'device' = 'mobile';
""", engine)

print("\n".join(plan["QUERY PLAN"]))
```

За допомогою `EXPLAIN (ANALYZE, BUFFERS)` було перевірено дві стратегії індексування на 50 000 синтетичних подій.

Спочатку запит із перевіркою входження (containment) для JSONB використовував послідовне сканування таблиці (Sequential Scan) і виконувався за 407,777 мс. Після створення GIN-індексу з `jsonb_path_ops` PostgreSQL почав використовувати `Bitmap Index Scan` і `Bitmap Heap Scan`, що скоротило час виконання до 7,151 мс — приблизно у 57 разів швидше.

Для виразу `payload ->> 'device' = 'mobile'` індекс B-tree на виразі скоротив час виконання з 51,692 мс до 3,517 мс, тобто приблизно у 14,7 раза.

GIN підходить для пошуку за структурами JSONB, тоді як індекси B-tree на виразах ефективні для поширених запитів за окремими скалярними значеннями, отриманими з JSONB.

Обидва індекси покращують продуктивність читання, але потребують додаткового місця для зберігання та створюють додаткове навантаження на обслуговування під час вставлення й оновлення даних. Фактична продуктивність залежить від вибірковості запиту, розподілу даних і кешування.

## Завдання 5. JSONB vs normalized + Reflection

```python
with engine.begin() as conn:
    conn.execute(text("""
        DROP TABLE IF EXISTS t7_hw_checkouts_norm;

        CREATE TABLE t7_hw_checkouts_norm (
            event_id        INTEGER PRIMARY KEY,
            user_id         INTEGER NOT NULL,
            event_ts        TIMESTAMPTZ NOT NULL,
            order_id        TEXT NOT NULL,
            total           NUMERIC(10, 2) NOT NULL,
            items_count     INTEGER NOT NULL CHECK (items_count > 0),
            payment_method  TEXT NOT NULL
                CHECK (payment_method IN ('card', 'paypal', 'crypto'))
        );

        INSERT INTO t7_hw_checkouts_norm (
            event_id,
            user_id,
            event_ts,
            order_id,
            total,
            items_count,
            payment_method
        )
        SELECT
            event_id,
            user_id,
            event_ts,
            payload ->> 'order_id',
            (payload ->> 'total')::NUMERIC(10, 2),
            (payload ->> 'items_count')::INTEGER,
            payload ->> 'payment_method'
        FROM t7_hw_events
        WHERE event_type = 'checkout';

        CREATE INDEX idx_t7_checkouts_norm_pm
            ON t7_hw_checkouts_norm (payment_method);

        ANALYZE t7_hw_checkouts_norm;
    """))

print("Normalized checkout table created successfully.")
```

```python
display(pd.read_sql("""
    SELECT
        COUNT(*) AS n_checkouts,
        COUNT(DISTINCT payment_method) AS n_payment_methods,
        MIN(total) AS min_total,
        MAX(total) AS max_total,
        AVG(total)::NUMERIC(10, 2) AS avg_total
    FROM t7_hw_checkouts_norm;
""", engine))
```

```python
plan_jsonb = pd.read_sql("""
    EXPLAIN (ANALYZE, BUFFERS)
    SELECT
        payload ->> 'payment_method' AS payment_method,
        AVG((payload ->> 'total')::NUMERIC) AS avg_total,
        COUNT(*) AS n
    FROM t7_hw_events
    WHERE event_type = 'checkout'
    GROUP BY payload ->> 'payment_method'
    ORDER BY avg_total DESC
    LIMIT 3;
""", engine)

print("\n".join(plan_jsonb["QUERY PLAN"]))
```

```python
plan_normalized = pd.read_sql("""
    EXPLAIN (ANALYZE, BUFFERS)
    SELECT
        payment_method,
        AVG(total) AS avg_total,
        COUNT(*) AS n
    FROM t7_hw_checkouts_norm
    GROUP BY payment_method
    ORDER BY avg_total DESC
    LIMIT 3;
""", engine)

print("\n".join(plan_normalized["QUERY PLAN"]))
```

```python
comparison = pd.read_sql("""
    WITH jsonb_result AS (
        SELECT
            payload ->> 'payment_method' AS payment_method,
            AVG((payload ->> 'total')::NUMERIC) AS avg_total,
            COUNT(*) AS n
        FROM t7_hw_events
        WHERE event_type = 'checkout'
        GROUP BY payload ->> 'payment_method'
    ),
    normalized_result AS (
        SELECT
            payment_method,
            AVG(total) AS avg_total,
            COUNT(*) AS n
        FROM t7_hw_checkouts_norm
        GROUP BY payment_method
    )
    SELECT
        j.payment_method,
        ROUND(j.avg_total, 2) AS jsonb_avg,
        ROUND(n.avg_total, 2) AS normalized_avg,
        j.n AS jsonb_count,
        n.n AS normalized_count,
        ROUND(j.avg_total, 2) = ROUND(n.avg_total, 2)
            AND j.n = n.n AS results_match
    FROM jsonb_result j
    JOIN normalized_result n
        ON n.payment_method = j.payment_method
    ORDER BY j.payment_method;
""", engine)

display(comparison)
```

### Reflection

У цій домашній роботі було показано, як PostgreSQL може підтримувати гнучке зберігання подій у форматі JSONB і використовувати різні стратегії індексування для підвищення продуктивності запитів.

До створення GIN-індексу запит із перевіркою входження даних у JSONB використовував послідовне сканування таблиці (`Seq Scan`) і виконувався за 407,777 мс. Під час цього сканувалися 50 000 подій, щоб знайти 3 334 відповідні записи. Після додавання GIN-індексу з `jsonb_path_ops` PostgreSQL почав використовувати `Bitmap Index Scan` і `Bitmap Heap Scan`, завдяки чому час виконання скоротився до 7,151 мс. Оціночна загальна вартість запиту зменшилася з 1761,48 до 1061,12.

GIN-індекси ефективні для перевірки входження даних у JSONB (`@>`), оскільки індексують структури JSON, за якими можна виконувати пошук. Однак запити на кшталт `payload ->> 'device' = 'mobile'` порівнюють текстові значення, отримані з JSONB. Для такого типу запитів краще підходить B-tree-індекс на виразі. У нашому експерименті він скоротив час виконання з 51,692 мс до 3,517 мс.

Такі покращення мають і свої компроміси. Індекси прискорюють читання даних, але потребують додаткового місця для зберігання та збільшують витрати на обслуговування під час вставлення й оновлення записів. Для GIN-індексу знадобилося приблизно 1 144 КБ дискового простору.

Для завантаження подій машинного навчання (ML) я б обрала JSONB, якщо різні типи подій мають різні атрибути або їхня структура може змінюватися з часом. Нормалізовані таблиці краще підходять для структурованих ознак, які часто використовуються, потребують суворої перевірки даних і мають ефективно агрегуватися. Наш запит до нормалізованих таблиць виконався за 7,373 мс, тоді як запит до JSONB — за 15,476 мс. При цьому результати обох запитів були однаковими.

Щоб додати новий тип події — повернення коштів (`refund`) — у JSONB, потрібно було б переважно визначити й перевірити нову структуру payload. У нормалізованому підході я б створила окрему таблицю для повернень коштів із типізованими стовпцями, обмеженнями та відповідними індексами. Це потребує додаткової роботи зі зміною схеми бази даних, але забезпечує надійніший контроль цілісності даних.

Загалом JSONB у PostgreSQL забезпечує гнучкість під час завантаження подій, тоді як нормалізовані структури краще підходять для підготовлених аналітичних даних і робочих навантажень машинного навчання.
